```python
import os, numpy as np, tqdm, pyrefCLASS as rot
from matplotlib.pyplot import *
```

```python
# initial condition and parameters for phi EoM
Neta = 10000
alpha0 = .3 # [deg]
n = 0 # maki
alpha = alpha0*np.pi/180. + n*np.pi # [rad]
```

```python
#mass = np.logspace(-28,-26,100)
mass = np.logspace(-27,-26,50)
```

```python
eta0 = rot.z2eta(0) # conformal time at present
etalist = eta0 * np.arange(1, Neta + 1) / Neta
```

```python
zlist = rot.eta2z(etalist)
```

```python
xlim(700,1500)
for m in tqdm.tqdm(mass[::5]):
    filename_dynamics = rot.filename_axion_dynamics(Neta,[m])
    phi = np.loadtxt('data_axion/'+filename_dynamics)
    plot(zlist,phi,label=str(np.log10(m)))
legend()
```
