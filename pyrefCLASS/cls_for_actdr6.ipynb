```python
import os, numpy as np, tqdm, pyrefCLASS
```

```python
# initial condition and parameters for phi EoM
Neta = 10000
alpha0 = .3 # [deg]
n = 0 # maki
```

```python
alpha = alpha0*np.pi/180. + n*np.pi # [rad]
```

```python
mass = np.concatenate( ( np.logspace(-27,-26,500), np.logspace(-27,-26,50), np.logspace(-28,-26,100) ) )
mass = np.unique(mass)
#mass = np.logspace(-28,-26,100)[::5]
```

```python
for m in tqdm.tqdm(mass,leave=False):
    if np.log10(m) < -26.45: continue
    print(m)
    filename_dynamics = pyrefCLASS.filename_axion_dynamics(Neta,[m])
    base_tag = 'alpha' + str(alpha0) + 'deg' + filename_dynamics.replace('axionDynamics', '').replace('.dat', '') + '+'
    #ofile = '../../data_local/2025-01-09_ALP/'+base_tag+'cl.dat'
    ofile = 'output/'+base_tag+'cl.dat'
    if os.path.exists(ofile):
        print(ofile+' exist')
        continue
    else:
        pyrefCLASS.export_axion_dynamics(Neta, pyrefCLASS.EoM_phi_mass_potential, [m])
        pyrefCLASS.create_inifile(alpha, base_tag, Neta, filename_dynamics, default_ini_file='biref_default_actdr6.ini')
        #! {pyrefCLASS.path_to_birefCLASS}/class ./initial_file/{base_tag}.ini {pyrefCLASS.path_to_birefCLASS}/cl_ref_biref.pre > test.txt
        #! {pyrefCLASS.path_to_birefCLASS}/class ./initial_file/{base_tag}.ini {pyrefCLASS.path_to_birefCLASS}/cl_ref_birefB.pre > test.txt
        ! {pyrefCLASS.path_to_birefCLASS}/class ./initial_file/{base_tag}.ini {pyrefCLASS.path_to_birefCLASS}/cl_ref_birefC.pre > test.txt
        #! {pyrefCLASS.path_to_birefCLASS}/class ./initial_file/{base_tag}.ini {pyrefCLASS.path_to_birefCLASS}/cl_ref_birefE.pre > test.txt
        for tag in ['cls.dat','clt.dat','parameters.ini','primordial_Pk.dat','thermodynamics.dat','unused_parameters','background.dat']:
            os.system('rm -rf output/'+base_tag+tag)
```
